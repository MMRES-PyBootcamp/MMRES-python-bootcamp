## Misophonia data set description

Here is a brief description of the variables present in the Misophonia data set:

* **Misophonia**: Binary variable (string) stating the presence or absence of the misophonia condition:
  - "yes": with misophonia
  - "no": without misophonia

.
* **Misophonia Severity**: Categorical (integer) variable describing misophonia condition severity:
  - 0: no misophonic
  - 1: low severity
  - 2: mid severity
  - 3: high severity
  - 4: very high severity

.
* **Marital Status**: Categorical variable (string) describing marital status:
  - "single"
  - "married"
  - "divorsed" (Note: Misspelled. Should say "divorced")
  - "widow" (Note: Misspelled. Should say "widowed")

.
* **Marital Status Numeric**: Categorical variable (integer) describing marital status:
  - 0: Single
  - 1: Married
  - 2: Divorced
  - 3: Widowed

.
* **Anxiety Trait**: Score (integer) from 0 to 100 describing the anxiety personality trait.

.
* **Axiety Trait Dichotomic**: Binary variable (integer) describing the anxiety personality trait (Note: Misspelled. Should say "Anxiety"):
  - 0: without anxiety
  - 1: with anxiety

.
* **Axiety State**: Score (integer) from 0 to 100 describing current anxiety state.

.
* **Anxiety State Dichotomic**: Binary variable (integer) describing current anxiety state:
  - 0: without anxiety
  - 1: with anxiety

.
* **Anxiety Medication**: Binary variable (string) stating the existence of a previously diagnosed and pharmacologically treated anxiety disorder:
  - "yes": with pharmacologically treated anxiety disorder
  - "no": without pharmacologically treated anxiety disorder

.
* **Depression**: Score (integer) from 0 to 50 describing current state of depression.

.
* **Depression Dichotomic**: Binary variable (integer) describing the current state of depression:

  - 0: without depression
  - 1: with depression

.
* **Sex**: Binary variable (string) describing the sex of the individual:
  - "female"
  - "male"

.
* **Age**: Numerical variable (integer) describing the age of the individual.

.
* **Jaw Type**: Categorical variable (string) describing the [jaw type](https://en.wikipedia.org/wiki/Jaw_abnormality):
  - "I": Class I (Normal / Neutral Occlusion)
  - "II": Class II (Retrognathism / Overbite)
  - "III": Class III (Prognathism / Underbite)

.
* **Convexity Angle**: Numerical variable (float) reporting the facial convexity angle (likely, in degrees).

.
* **Mandibular Protrussion**: Numerical variable (float) reporting the mandibular protrusion (likely, in millimeters) (Note: Misspelled. Should say "Protrusion").

.
* **Neck and Lower 3rd Angle**: Numerical variable (float) reporting the facial "neck and Lower third angle (likely, in degrees)"

.
* **Subnasal Angle**: Numerical variable (float) reporting the facial subnasal angle (likely, in degrees).

.
* **Self Perception Change**: Numerical variable (integer, but expressed as float) reporting whether people changed (5) their self-concept after misophonia-specific treatment or not (0).

.
* **Misophonia Post**: Numerical variable (float) reporting the A-MISO score (obtained during misophonia diagnosis), after an educational program, where patients were made aware of a condition called misophonia.

.
* **Misophonia Pre**: Numerical variable (float) reporting the A-MISO score (obtained during misophonia diagnosis), before an educational program, where patients were made aware of a condition called misophonia.

## Scoring criteria

Keep in mind that the main objective of the Misophonia data set analysis is to evaluate if you have acquired the basic skills needed to analyze a generic data set using Python. Partial and final results can be statistically significant or not. We are interested in **meaningful results** in the context of your particular analyses with a proper **justification** and **interpretation**.

Since the Misophonia analysis is completely open, providing a deterministic evaluation is complex, if not impossible. With the aim of better guiding you in the data set analysis and Flash Talk preparation, we list below some key points that we consider important to assign the final grade:

* **Timing (10%)**: Please, strictly stick to the nominal 5' minutes slot to maximize evaluation scoring. As you can see, the time available is very limited.

* **Sharing (10%)**: Please, try to evenly distribute the presentation time between you and your couple partner, so both have the opportunity to expose a part of the job and be properly evaluated (this also applies to the Q&A part). 

* **Formalism (20%)**: Despite the final exam will cover the evaluation of the SDA theory block, we obviously expect that the strategy and the methods deployed to statistically analyze being formally correct.

* **Structuring (25%)**: A good structure enables a clear exposition. We are not looking for the classical structure (abstract, introduction, methods, results, discussion, conclusions). We just want to see a gradual and logical construction of your analysis. As we said, 1) hypothesis exposition, 2) data analysis strategy and methods, 3) results and Interpretation is an example of what we could consider a nice structure (but maybe not the unique one, there is always room for creativity...).

* **Plotting (25%)**: The plots you present should meet the highest scientific quality standards. In particular:

  + The magnitudes represented should be properly labeled, indicating their units if necessary. Remember that plot legends are also mandatory.
 
  + Axis labels, tick labels, figure title, figure captions, etc... should be easily readable and visible.

  + Depending on the particular characteristics of your data and depending on where you want to put focus, one visualization could be more suitable than other. Choose the visualization strategy that better represents the insights you want to highlight.
 
*  **Chairing (10%)** : In addition to evaluating your Flash Talk, we will also evaluate your chairing. While chairing, try to identify and clearly express the main strength of your peer's presentation. Formulate a clear and relevant scientific question / improvement / clarification (always in the most constructive way possible). Put in practice your respect, active listening, and critical thinking capabilities.
